```python
# import the graph
import json

CN_GRAPH = json.load(open('../bank_files/full_word_graph_noweight.json'))
```

```python
# sort the keys of the graph by the number of entries
sorted_keys = sorted(CN_GRAPH.keys(), key=lambda x: sum(len(entry) for entry in CN_GRAPH[x]), reverse=True)
```

```python
print("Sorted keys by number of entries:")
for key in sorted_keys[0:40]:
    print(f"{key}: {sum(len(entry) for entry in CN_GRAPH[key])}")
```
